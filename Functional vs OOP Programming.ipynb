Sort an Array of 0s, 1s, and 2s

```python

class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1

        return arr


arr = [0, 2, 1, 2, 0, 1]
solution = Solution()

print("Original array:", arr)
print("Sorted array:", solution.sort012(arr.copy()))
```

Binary Search

```python

class Solution:
    def binarysearch(self, arr, k):
        left = 0
        right = len(arr) - 1

        while left <= right:
            mid = (left + right) // 2

            if arr[mid] == k:
                return mid
            elif arr[mid] < k:
                left = mid + 1
            else:
                right = mid - 1

        return -1


numbers = [1, 3, 5, 7, 9, 11]
target = 7

solution = Solution()
result = solution.binarysearch(numbers, target)

print("Array:", numbers)
print("Target:", target)
print("Index:", result)
```
